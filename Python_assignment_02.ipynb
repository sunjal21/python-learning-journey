#4. String Manipulation:

occurrences of a specific character using a for loop.

Input (char which you want to find)- x
Output - x , 0

Step 1: define a string

Step 2: enter character which you want to search

Step 3: initialize count to 0 i.e count=0

Step 4: use for loop

Step 5: use if condition inside for loop to check if given character matches with the string element
e.g if i== character

```python
str1 = input("Enter a string: ")
character = input("Enter a character to find: ")

count = 0

for i in str1:
    if i == character:
        count = count + 1

print(count)
```

#5. Dictionary Processing:
 Implement a script that iterates through a dictionary and prints
both keys and values using a for loop.

Input- d = {“name” : “Amit” , “Age”:20}

Output 1 - [“name” , “age”]

Output 2 - [“Amit” , 20 ]

```python
d = {"name": "amit", "age": 20}
keys = []
values = []

for key in d.keys():
    keys.append(key)

for value in d.values():
    values.append(value)

print(keys)
print(values)
```

#8. Find Even Numbers:
 Develop a program that identifies and prints all even numbers
within a specified range using while loop.

Step-01 Define the range of integers and store in variables (named start , end)

Step-02 Start a while loop on variable named end

Step-03 Use % operator to check that whether the current number ( current_value) is
even or not , if number is even then print that number

Steo-04 Increment current_value by 1

Step-05 : Repeat Step-02 and

Step-04 until the while condition is True

```python
start = 1
end = 8

current_value = start

while current_value <= end:
    if current_value % 2 == 0:
        print(current_value)

    current_value += 1
```

#10. Dictionary Processing:
Implement a script that iterates through a dictionary and prints
both keys and values using a while loop.

```python
dic1 = {"name": "sunjal", "age": 20}

keys = []
values = []

items = list(dic1.items())
i = 0

while i < len(items):
    keys.append(items[i][0])
    values.append(items[i][1])
    i += 1

print(keys)
print(values)
```
