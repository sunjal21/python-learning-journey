### Use Case 1: Validating Email Addresses

**Scenario:** You need to validate user input to ensure it's a properly formatted email address (e.g., `user@example.com`).

**Task:** Create a RegEx to validate a common email format.

```python
import re
```

```python
test_sample='''Please review the following list of email registrations to verify their formats:

VALID ACCOUNTS:
- Regular user: john.doe@example.com
- Corporate style: alice_smith123@company.co.uk
- Tech support with subdomain: support@mail.dev-hub.io
- Marketing tagging: newsletter+promo2026@marketing-agency.net
- Minimalist format: a@b.org

INVALID ACCOUNTS (Should be ignored or fail validation):
- Missing domain extension: test@com
- Missing domain name: user@.com
- Missing at symbol: ://domain.com
- Double at symbol: double@@example.com
- Spaces in local part: spaces are bad@domain.com
- Trailing dot: user@example.com.
- Starting with a dot: .invalid@domain.com'''

test_pattern=r'[\w]+@[\w]+\.[\w]+'

re.findall(test_pattern,test_sample)
```

### Use Case 2: Extracting Phone Numbers

**Scenario:** You have a block of text and need to extract all phone numbers that follow a specific format (e.g., `(XXX) XXX-XXXX` or `XXX-XXX-XXXX`).

**Task:** Create a RegEx to find and extract phone numbers in these common formats.

```python
import re

Test_phase = 'my number is 129-456-7890'

print(f"your number is {re.findall(r'\([0-9]{3}\)-[0-9]{3}-[0-9]{4}|[0-9]{3}-[0-9]{3}-[0-9]{4}', Test_phase)}")
```

#OR

```python
test_1='123-456-7890'
print(f"your number is {re.findall(r'\(\d{3}\)-\d{3}-\d{4}|\d{3}-\d{3}-\d{4}',test_1)}")
```

### Use Case 3: Parsing Log Files for Error Codes

**Scenario:** You are monitoring server logs and need to identify specific error codes or messages, such as `ERROR: [CODE-123]` or `WARN: [HTTP-500]`.

**Task:** Develop a RegEx to capture the error/warning level and the code within the brackets.

```python
text = '''2026-09-03 14:22:01 INFO: [SYS-100] System initialization started successfully.
2026-09-03 14:23:15 ERROR: [CODE-123] Database connection timeout after 30 seconds.
2026-09-03 14:23:45 WARN: [HTTP-500] Internal server error received from authentication service.
2026-09-03 14:24:12 DEBUG: [SQL-042] Executing query: SELECT * FROM users WHERE active = 1;
2026-09-03 14:25:00 ERROR: [AUTH-401] Unauthorized access attempt detected from IP 192.168.1.50.
2026-09-03 14:26:30 CRITICAL: [HARD-999] Hard drive disk space running critically low (less than 1%).
2026-09-03 14:27:11 WARN:[MIS-FORMAT] Missing space after colon should not be matched.
2026-09-03 14:28:05 ERROR: CODE-456 Missing brackets around the code entirely.
'''
pattern = r'(ERROR|WARN): \[([A-Z-0-9]+)\]'
re.findall(pattern,text)
```

### Use Case 4: Validating Passwords

**Scenario:** A website requires passwords to meet certain criteria: at least 8 characters long, containing at least one uppercase letter, one lowercase letter, one digit, and one special character.

**Task:** Write a RegEx to validate a password against these requirements.


```python
test_phase_2="Abcd@123"

test_case_3=  r'[A-Z]+','[a-z]+','[!@#$%^&*]','d+'

if len(test_phase_2)>=8 and all (re.search(pattern,test_phase_2) for pattern in test_case_3):
  print("Password is strong")
else:
  print("Password is weak, retry")
```

*NOTE*  -  **.** means any character except newline by default.

```python
text = '''Password123!
abc
SecretPass!@2026
Valid_Pass99
weakpassword
ONLYUPPERCASE123
1234567890!@#
NoSpecialChar#9'''
pattern=r'(?=.*[A-Z]+)(?=.*[a-z]+)(?=.*[!@#$%^&_*])(?=.*[0-9]).{8,}'
re.findall(pattern,text)
```

### Use Case 5: Extracting URLs from Text

**Scenario:** You have a document and need to find all URLs (http or https) embedded within the text.

**Task:** Create a RegEx to identify and extract full URLs (e.g., `https://www.google.com` or `http://example.org/path/to/page.html`).

```python
import re

text = f"""Visit https://www.google.com or http://example.org/path/to/page.html for more information."""

pattern = r'https?://[^\s]+'
re.findall(pattern,text)
```
